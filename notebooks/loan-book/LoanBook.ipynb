# Loan Book Diversification

Monte Carlo simulation of independent loan defaults (2% PD, 40% LGD) to show how portfolio loss risk falls as the book gets larger.


```python
import numpy as np
```

```python
rng = np.random.default_rng() # Random number function, between 0 and 1
defaults = rng.random((5000, 20000)) < 0.02
```

```python
print(defaults)
print(defaults.mean()) # The rate of loans which default, 2%
```

```python
loss = defaults * 0.4 # Lose 40% of the loan when it defaults
YearLoss = loss.mean(axis=1) # Each row is one year, and the YearLoss is what fraction of the book was lost that year
print(YearLoss)
    
```

```python
sizes = np.array([100, 1000, 20000])
typical = np.empty(3) # typical year for book size
bad = np.empty(3) # bad year for book size

for i, n in enumerate(sizes): # i is the number loop number, n is the size of each book (from sizes array)
    defaults = rng.random((5000, n)) < 0.02
    loss = defaults * 0.4
    YearLoss = loss.mean(axis=1)
    typical[i] = YearLoss.mean()
    bad[i] = np.quantile(YearLoss, 0.95)

print("loans   typical     bad year")
for i in range(3):
    print(f"{sizes[i]:<8}{typical[i] * 100:.2f}%       {bad[i] * 100:.2f}%")
```

Larger books have similar *average* losses, but much thinner left-tail (bad-year) losses — diversification of idiosyncratic credit risk.
