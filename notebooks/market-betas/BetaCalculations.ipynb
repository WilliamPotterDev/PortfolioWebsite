# Equity Beta Estimation

Estimate equity betas for major tech names against the CRSP value-weighted market return:

$$\beta_i = \frac{\mathrm{Cov}(R_i, R_m)}{\mathrm{Var}(R_m)}$$


```python
import pandas as pd

dataframe = pd.read_excel('crsp2025.xlsx')
```

```python
# Covariance of Daily Total Return and Value-weighted return including dividends / variance of Value weighted return including dividends
aapl = dataframe[dataframe['Ticker'] == 'AAPL']
aaplBeta = aapl['Daily Total Return'].cov(aapl['Value-weighted return including dividends']) / aapl['Value-weighted return including dividends'].var()
```

```python
nvda = dataframe[dataframe['Ticker'] == 'NVDA']
nvdaBeta = nvda['Daily Total Return'].cov(nvda['Value-weighted return including dividends']) / nvda['Value-weighted return including dividends'].var()
```

```python
avgo = dataframe[dataframe['Ticker'] == 'AVGO']
avgoBeta = avgo['Daily Total Return'].cov(avgo['Value-weighted return including dividends']) / avgo['Value-weighted return including dividends'].var()
```

```python
msft = dataframe[dataframe['Ticker'] == 'MSFT']
msftBeta = msft['Daily Total Return'].cov(msft['Value-weighted return including dividends']) / msft['Value-weighted return including dividends'].var()
```

```python
orcl = dataframe[dataframe['Ticker'] == 'ORCL']
orclBeta = orcl['Daily Total Return'].cov(orcl['Value-weighted return including dividends']) / orcl['Value-weighted return including dividends'].var()
```

```python
print(f'{'    Apple Beta:':<20}{aaplBeta}')
print(f'{'   Nvidia Beta:':<20}{nvdaBeta}')
print(f'{' Broadcom Beta:':<20}{avgoBeta}')
print(f'{'Microsoft Beta:':<20}{msftBeta}')
print(f'{'   Oracle Beta:':<20}{orclBeta}')
```
